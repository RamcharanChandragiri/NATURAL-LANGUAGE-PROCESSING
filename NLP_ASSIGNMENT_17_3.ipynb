```python
from transformers import pipeline
import pandas as pd
```

```python
df = pd.DataFrame({
    "English": [
        "The weather is pleasant today.",
        "I am going to college.",
        "How are you?",
        "Artificial intelligence is transforming industries.",
        "Can you help me with this task?",
        "The train arrived late.",
        "She completed her assignment.",
        "Machine learning requires data.",
        "The meeting was postponed.",
        "Do you like coffee?",
        "Natural language processing is important.",
        "Although it was raining, we went outside.",
        "The exam was difficult.",
        "Please send the report tomorrow.",
        "If you study hard, you will succeed."
    ]
})

df
```

```python
sentences = df["English"].tolist()

for i, sentence in enumerate(sentences, 1):
    print(f"{i}. {sentence}")
```

```python
translator_fr = pipeline(
    "translation_en_to_fr",
    model="Helsinki-NLP/opus-mt-en-fr"
)

translator_de = pipeline(
    "translation_en_to_de",
    model="Helsinki-NLP/opus-mt-en-de"
)

print("French and German models loaded successfully")
```

```python
print("French:", translator_fr("How are you?")[0]["translation_text"])
print("German:", translator_de("How are you?")[0]["translation_text"])
```

```python
translated_french = []
translated_german = []

for sentence in sentences:
    fr = translator_fr(sentence)[0]["translation_text"]
    de = translator_de(sentence)[0]["translation_text"]

    translated_french.append(fr)
    translated_german.append(de)
```

```python
result_df = pd.DataFrame({
    "English": sentences,
    "French": translated_french,
    "German": translated_german
})

result_df
```

```python
print("TRANSLATION QUALITY ANALYSIS")
print("-" * 40)
print("1. Meaning is preserved in most sentences.")
print("2. Grammar is correct in both languages.")
print("3. Technical terms are translated well.")
print("4. German changes sentence structure slightly.")
print("5. Complex sentences remain understandable.")
print("6. Fluency is high in both outputs.")
```

```python
complex_sentences = [
    "Although the weather was bad, we continued the journey.",
    "Can you explain how machine learning works?",
    "The project was delayed because resources were unavailable.",
    "If you practice daily, you will improve.",
    "Artificial intelligence is changing modern technology."
]

for sentence in complex_sentences:
    print("English :", sentence)
    print("French  :", translator_fr(sentence)[0]["translation_text"])
    print("German  :", translator_de(sentence)[0]["translation_text"])
    print("-" * 80)
```
