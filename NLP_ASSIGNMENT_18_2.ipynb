```python
from transformers import pipeline

summarizer = pipeline("summarization", model="facebook/bart-large-cnn")

print("Model loaded successfully!")
```

```python
texts = [
"""Artificial Intelligence (AI) is rapidly transforming industries worldwide. It enables machines to learn from data, recognize patterns, and make decisions. AI is used in healthcare, finance, education, and transportation. However, ethical concerns such as job loss and data privacy remain challenges.""",

"""The new smartphone launch introduced innovative features such as advanced cameras, faster processors, and improved battery life. Consumers are excited, but high prices remain a concern for many users.""",

"""In academic research, data analysis plays a crucial role in drawing conclusions. Researchers use statistical tools to interpret data and validate hypotheses. Accuracy and reliability are essential.""",

"""Software development includes planning, coding, testing, and maintenance. Developers must ensure software is efficient, secure, and user-friendly. Agile methodology is commonly used.""",

"""Once upon a time, there was a small village surrounded by mountains. The villagers lived peacefully and depended on farming. One day, a traveler arrived and changed their lives.""",

"""Person A: How was your day? Person B: It was busy but productive. I had meetings and completed my work. Person A: That sounds great!"""
]
```

```python
for i, text in enumerate(texts):
    summary = summarizer(text, max_length=100, min_length=30, do_sample=False)
    print(f"\nText {i+1} Summary:\n{summary[0]['summary_text']}")
```

```python
text = texts[0]

short_summary = summarizer(text, max_length=40, min_length=15, do_sample=False)
medium_summary = summarizer(text, max_length=80, min_length=25, do_sample=False)
long_summary = summarizer(text, max_length=120, min_length=40, do_sample=False)

print("Short Summary:\n", short_summary[0]['summary_text'])
print("\nMedium Summary:\n", medium_summary[0]['summary_text'])
print("\nLong Summary:\n", long_summary[0]['summary_text'])
```
